# Тести для малих вибірок та пропорцій



**Завдання 1**. E-commerce компанія після редизайну сайту підозрює, що **середній час до покупки (time-to-purchase)** користувачів **збільшився**.
Історично середній час від першого заходу на сайт до покупки становив $\mu_0 = 123.7$ хвилин.

Команда аналітиків випадково обрала дані **7 покупців після редизайну**:
`128, 135, 121, 142, 126, 151, 123`

З допомогою наявних даних зʼясуйте, чи збільшився середній час покупки після редизайну.

Для цього виконайте наступні 6 кроків. Правильне виконання кожного з кроків оцінюється в 1 бал.

1. Запишіть нульову та альтернативну гіпотези і визначте тип тесту.
2. Обчисліть вибіркові статистики: $\bar x$, $s$, $n$.
3. Оберіть тип тесту та виконайте його (знайдіть критичне значення тесту, статистику тесту та р-значення) будь-яким пасуючим способом, щоб перевірити гіпотезу на рівні значущості $\alpha = 0.10$.
4. Прийміть рішення, чи відхиляєте ви гіпотезу $H_0$ використовуючи p-value.
5. Напишіть висновок: чи справді редизайн сайту подовжив час до покупки?
6. Чи зміниться ваше рішення при зміні рівня значущості на $\alpha = 0.05$.


```python
import numpy as np
from scipy import stats

# Дані з умови: час до покупки після редизайну, хв
time_to_purchase = np.array([128, 135, 121, 142, 126, 151, 123])
mu0 = 123.7   # історичний середній час до покупки, хв
alpha = 0.10

# Крок 1. Гіпотези
# H0: μ = 123.7
# H1: μ > 123.7
# Тип тесту: правосторонній

# Крок 2. Вибіркові статистики
x_bar = np.mean(time_to_purchase)
s = np.std(time_to_purchase, ddof=1)   # стандартне відхилення з n-1
n = len(time_to_purchase)

# Крок 3. t-статистика вручну
t_stat = (x_bar - mu0) / (s / np.sqrt(n))

# Критичне значення для правостороннього тесту
df = n - 1
t_crit = stats.t.ppf(1 - alpha, df)

# p-value (правий хвіст)
p_value = stats.t.sf(t_stat, df)

print(f"x̄ = {x_bar:.3f}, s = {s:.3f}, n = {n}")
print(f"t = {t_stat:.4f}, t_crit = {t_crit:.4f}, p-value = {p_value:.4f}")
```

Крок 4. p-value = 0.042 < α = 0.10, тому H0 відхиляємо

Крок 5. На рівні значущості 0.10 є статистично значущі докази, що середній час до покупки після редизайну збільшився. Вибіркове середнє 132.3 хв перевищує історичні 123.7 хв приблизно на 8.6 хв. Вибірка дуже мала (7 покупців), тому висновок варто підтвердити на більших даних.

```python
alpha = 0.05

# Критичне значення для правостороннього тесту
df = n - 1
t_crit = stats.t.ppf(1 - alpha, df)

# p-value (правий хвіст)
p_value = stats.t.sf(t_stat, df)

print(f"t_crit = {t_crit:.4f}, p-value = {p_value:.4f}")
```

Крок 6. t = 2.069 > 1.943, тож потрапляє в критичну область.
p-value = 0.042 < 0.05. Рішення не змінюється: H0 відхиляємо і при α = 0.05, хоч p-value лише трохи нижче порогу.

**Завдання 2.**

До спеціальної рекламної кампанії **23%** дорослих упізнавали логотип компанії. Після завершення кампанії відділ маркетингу провів опитування: з **1200** випадково відібраних дорослих **311** упізнали логотип.

Перевірте на рівні значущості **3%** ($\alpha=0.03$), чи дають ці дані достатні підстави стверджувати, що **тепер більше ніж 23%** дорослих упізнають логотип компанії. Для розвʼязку використовуйте бібліотеку `statsmodels`.

Зробіть висновок, чи зросла впізнаваність логотипу.

Додатково, обчисліть довірчий інтревал на заданому рівні значущості і проінтерпретуйте текстом - як він додатково пояснює прийняте нами рішення?

```python
import numpy as np
from statsmodels.stats.proportion import proportions_ztest, proportion_confint

# Дані з умови
count = 311      # упізнали логотип
nobs = 1200      # розмір вибірки
p0 = 0.23        # частка до кампанії
alpha = 0.03     # рівень значущості

# Гіпотези
# H0: p = 0.23
# H1: p > 0.23
# Тип тесту: правосторонній

# z-тест для частки
z_stat, p_value = proportions_ztest(count, nobs, value=p0,
                                    alternative='larger', prop_var=p0)

print(f"p̂ = {count / nobs:.4f}")
print(f"z = {z_stat:.4f}, p-value = {p_value:.4f}")

if p_value < alpha:
    print("p-value < α → відхиляємо H0: впізнаваність логотипу зросла")
else:
    print("p-value ≥ α → не відхиляємо H0: підстав для висновку про зростання недостатньо")

# Двосторонній 97% довірчий інтервал (alpha = 0.03)
ci_low, ci_high = proportion_confint(count, nobs, alpha=alpha, method='normal')
print(f"97% ДІ: [{ci_low:.4f}; {ci_high:.4f}]")

# Одностороння нижня межа (відповідає правосторонньому тесту)
ci_low_one, _ = proportion_confint(count, nobs, alpha=2 * alpha, method='normal')
print(f"Одностороння нижня межа (97%): {ci_low_one:.4f}")
```
