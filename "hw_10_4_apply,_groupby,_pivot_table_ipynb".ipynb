1. Дослідимо, який філіал супермаркету ('Branch') є найприбутковішим. Для цього знайдіть сумарний прибуток за кожним філіалом і виявіть, який філіал має найвищий.

```python
profit_by_branch = df.groupby('Branch')['gross income'].sum()

print(profit_by_branch)

print("Найприбутковіший філіал:", profit_by_branch.idxmax())
```

2. В якому місті знайходиться філіал з найвищим прибутком? Може в тому місці нам розмітисти ще один магазин.  
Знайдіть відповідь за допомогою функціоналу Pandas.

```python
most_profitable_branch = profit_by_branch.idxmax()

city = df.loc[df['Branch'] == most_profitable_branch, 'City'].iloc[0]

print(city)
```

3.1. Створіть зводну таблицю, яка покаже, скільки покупок (інвойсів) було зроблено в кожній з філій (`Branch`) за різними категоріями товарів. Запишіть таблицю в змінну `invoices_by_category` і виведіть змінну на екран.
Ця таблиця допоможе проаналізувати, в якій філії купують найбільше товарів кожної з категорій.

```python
invoices_by_category = pd.pivot_table(
    df,
    index='Branch',
    columns='Product line',
    values='Invoice ID',
    aggfunc='count'
)

invoices_by_category
```

Які категоріі товарів приносять найбільший середній чек з однієї покупки?

```python
avg_profit_by_category = df.groupby('Product line')['gross income'].mean().sort_values(ascending=False)

print(avg_profit_by_category)
```

Найвищий середній прибуток приносить категорія Home and lifestyle — 16.03, можна провести рекламні заходи для збільшення попиту.Найнижчий — Fashion accessories — 14.53 - варто проаналізувати ціни, місцерозташування, можливо винести до каси, чи змінити викладку.

Які категорії товарів жінки та чоловіки купують частіше — чи справді жінки тягнуться до дорожчих категорій, чи просто купують частіше загалом?

```python
df.groupby('Gender')['Product line'].value_counts(normalize=True).unstack().round(3) * 100
```

```python
df.groupby(['Product line', 'Gender'])['Total'].mean().round(2).unstack()
```

Загальна перевага жінок у середньому чеку тримається переважно на двох категоріях (їжа/напої та товари для дому), а не на всьому асортименті одразу.
